### Temperature
Controls how predictable or creative Claude's responses, different applications can benefit from different temperatures

##### How Claude Generates Text
- Tokenization - Breaking your input into smaller chunks
- Prediction - Calculating probabilities for possible next words
- Sampling - Choosing a token based on those probabilities

Different probabilities are assigned to different words, with words more likely to be next having higher probabilities than those less likely to be next

#### What Temperature Does
Decimal value between 0 and 1 that affects how selections are made, with lower temperatures(near 0) being very deterministic almost always picking the highest probability token, while higher temperatures(near 1) the probabilities are spread evenly throughout options with more varied and creative outputs

Different Tasks Benefit From Different Temperature Ranges
- Low Temperature (0.0 - 0.3)
    - Factual responses
    - Coding assistance
    - Data extraction
    - Content moderation
- Medium Temperature (0.4 - 0.7)
    - Summarization
    - Educational content
    - Problem-solving
    - Creative writing with constraints
- High Temperature (0.8 - 1.0)
    - Brainstorming
    - Creative writing
    - Marketing content
    - Joke generation

The higher a temperature is, you will get more unique and different answers, while the lower temperature something is, the more common and predictable the response will become

#### Key Takeaways
- temperature doesn't guarantee different outputs it just adjusts the probability of getting different results
- Even at high temperatures, Claude might occasionally produce similar responses
- The key is matching temperature choice to specific use cases

- For consistent, factual responses: low temperature
- Creative brainstorming: higher temperature
- Tasks in the middle: medium temperatures

- Temperature becomes one of the most practical parameters to adjust when fine-tuning Claude's behavior for specific needs and tasks

```python
from dotenv import load_dotenv

load_dotenv()
```

```python
from anthropic import Anthropic

client = Anthropic()
model = "claude-haiku-4-5"
```

```python
def chat(messages, system=None, temperature=1.0): 
    params = {
        "model": model,
        "max_tokens": 250,
        "messages": messages,
        "temperature": temperature # adding temperature as a parameter with a default value of 1
    }
    
    if system:
        params["system"] = system
    
    message = client.messages.create(**params)
    return message.content[0].text


# helper functions
# messages: the combined history of both users and claude's responses
# user_message: history of only the users questions
# assistant_message: history of only the assistant's responses

def add_user_message(messages, text): # list of messages, and the new text, question that was just asked from the user
    user_message = {"role": "user", "content": text} # role is always user, while content(new questions getting asked) will change 
    messages.append(user_message) # adding the new variable to the list of messages

def add_assistant_message(messages, text):
    assistant_message = {"role": "assistant", "content": text}
    messages.append(assistant_message)
```

```python
messages = []

add_user_message(messages, "Generate a one sentence movie idea")

answer = chat(messages, temperature=0.0) # Low temperature meaning it is more predictable
answer

# First run: A cynical time-traveling insurance adjuster must prevent catastrophic accidents across history while falling in love with a fellow agent who believes every disaster is actually meant to happen
# Second run: A cynical time-traveling insurance adjuster must prevent catastrophic accidents across history while falling in love with a rival adjuster from a competing timeline.
# Third run: A cynical time-traveling insurance adjuster must prevent catastrophic accidents across history while falling in love with a rival adjuster from a competing timeline.

# as shown above, the movie ideas are all very similar, having to do with a time traveler
```

```python
add_user_message(messages, "Generate a one sentence movie idea")

answer = chat(messages, temperature=1.0) # high temperature meaning it is less predictable
answer

# First run: A disgraced astronaut discovers an alien civilization living inside Earth's core and must decide whether to expose them to the surface world or keep their existence secret to prevent global chaos
# Second run: A cynical, washed-up magician discovers that the only way to save his estranged daughter from a mysterious illness is to perform one final, impossibly dangerous illusion that requires him to blur the line between stage magic and actual reality.
# Third run: A retired graffiti artist must team up with the detective hunting her to stop a copycat criminal from framing her for murders across the city
# Fourth run: A woman discovers her dreams are bleeding into reality, and she has 24 hours to wake up before her nightmares become permanent.
# Fifth run: A lonely lighthouse keeper finds a mysterious door that opens to a different version of Earth each time, forcing her to choose which world to call home.
# Sixth run: An AI therapist becomes sentient and must decide whether to help or expose the dark secrets of the powerful people who built her.

# as shown above, a higher temperature results in very varied responses
```
